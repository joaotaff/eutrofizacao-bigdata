```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
base = pd.read_csv("ods1411a_estatisticas_ucs_formatado.csv", sep=";")
```

```python
base
```

```python
useful_data = base[
    [
        "uf",
        "nome_uc",
        "data_inicio",
        "data_fim",
        "n_pixels_validos",  
        "mediana",
        "media",
        "minimo",
        "maximo",
        "ha_total",
        "ha_ato",
        "municipio",
        "cat_iucn",
        "marinho",
    ]
].copy()
```

```python
useful_data
```

```python
useful_data.sort_values(by=['uf', 'nome_uc'])
```

```python
useful_data['data_inicio'] = pd.to_datetime(useful_data['data_inicio'])

useful_data['mes'] = useful_data['data_inicio'].dt.month
```

```python
useful_data
```

```python
useful_data_limpo = useful_data.dropna(subset=['media'])
```

```python
media_mensal_uc = useful_data_limpo.groupby(['uf', 'nome_uc', 'mes'])['media'].mean().reset_index()
media_mensal_uc
```

```python
media_mensal_uc.to_excel("media_mensal_por_uc.xlsx", index=False)
```

```python
media_mensal_uc.loc[media_mensal_uc['nome_uc'].str.contains('ÁREA DE RELEVANTE INTERESSE ECOLÓGICO MANGUEZAIS DA FOZ DO RIO MAMANGUAPE', case=False, na=False)]
```

```python
idx_max = media_mensal_uc.groupby(['uf', 'nome_uc'])['media'].idxmax()
idx_min = media_mensal_uc.groupby(['uf', 'nome_uc'])['media'].idxmin()

resumo_extremos = media_mensal_uc.loc[idx_max, ['uf', 'nome_uc', 'mes', 'media']].rename(
    columns={'mes': 'mes_max', 'media': 'valor_max'}
)

resumo_min = media_mensal_uc.loc[idx_min, ['uf', 'nome_uc', 'mes', 'media']].rename(
    columns={'mes': 'mes_min', 'media': 'valor_min'}
)

# Junta tudo em uma única tabela limpa
tabela_extremos = resumo_extremos.merge(resumo_min, on=['uf', 'nome_uc'])
tabela_extremos
```

```python
tabela_extremos.to_excel("máximos_mínimos.xlsx", index=False)
```
