```python
import pandas as pd
```

```python
base = pd.read_csv("ods1411a_estatisticas_ucs_formatado.csv", sep=";")
```

```python
base.columns
```

```python
nome_uc = base["nome_uc"].copy().unique()
nome_uc
```

```python
matriz = base.groupby('nome_uc', as_index=False).agg({
    'ha_total': 'first',      # Use 'first' se a área já for o valor total repetido
    'esfera': 'first',
    'uf': 'first'
})
```

```python
matriz
```

```python
matriz.to_excel("unidades_conservacao.xlsx", index=False)
```
