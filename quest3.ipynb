```python
dados_spespa['data_inicio'] = pd.to_datetime(dados_spespa['data_inicio'])
dados_trindade['data_inicio'] = pd.to_datetime(dados_trindade['data_inicio'])

plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

plt.scatter(dados_spespa['data_inicio'], dados_spespa['media'], alpha=0.6, label='São Pedro e São Paulo', color='#1f77b4', s=25)
plt.scatter(dados_trindade['data_inicio'], dados_trindade['media'], alpha=0.6, label='Trindade e Martim Vaz', color='#ff7f0e', s=25)

plt.title('Série Temporal e Dispersão (Scatter Plot) - Média da Clorofila-a', fontsize=14, fontweight='bold')
plt.xlabel('Data de Início (Janela de 8 dias)', fontsize=12)
plt.ylabel('Média da Concentração de Clorofila-a ($mg/m^3$)', fontsize=12)
plt.legend(title='Unidade de Conservação')
plt.tight_layout()
plt.savefig('scatter_temporal_ucs.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

dados_combinados = pd.concat([
    dados_spespa.assign(UC="São Pedro e São Paulo"),
    dados_trindade.assign(UC="Trindade e Martim Vaz"),
])

sns.boxplot(
    x="UC",
    y="media",
    data=dados_combinados,
    palette=["#1f77b4", "#ff7f0e"],
    width=0.4,
)

plt.title(
    "Boxplot - Média da Clorofila-a",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Unidade de Conservação", fontsize=12)
plt.ylabel("Média da Concentração de Clorofila-a ($mg/m^3$)", fontsize=12)

plt.tight_layout()
plt.savefig("boxplot_comparativo_ucs.png", dpi=300)
plt.show()
```

```python
df_merged = pd.merge(
    dados_spespa[["data_inicio", "data_fim", "media"]],
    dados_trindade[["data_inicio", "data_fim", "media"]],
    on=["data_inicio", "data_fim"],
    suffixes=("_spespa", "_trindade")
)
df_merged
```

```python
matriz_cov = df_merged[["media_spespa", "media_trindade"]].cov()
print("Matriz de Covariância:")
print(matriz_cov)
```

```python
matriz_corr = df_merged[["media_spespa", "media_trindade"]].corr()
print("\nMatriz de Correlação (Pearson):")
print(matriz_corr)
```

```python
plt.figure(figsize=(9, 5))
sns.regplot(data=df_merged, x="media_spespa", y="media_trindade", color="tab:blue", line_kws={"color": "red"})
plt.title("Relação entre as Médias: São Pedro e São Paulo vs. Trindade")
plt.xlabel("Média - São Pedro e São Paulo")
plt.ylabel("Média - Trindade e Martim Vaz")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
